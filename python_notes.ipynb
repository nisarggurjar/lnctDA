### Slicing

`value[start:stop:step]` takes items from `start` up to, and not including, `stop`. Omit `start` to begin at the front. Omit `stop` to continue to the end. A step of `-1` reverses the sequence. The same rules work for strings and lists.


````python
x = "Hello Python"
print(x[4:9])
print(x[6:])
print(x[:9])

y = [1, 2, 3, 4, 5, 6, 7, 8, 9]
print(y[3:8])
print(y[4:-2])
print(y[0:8:2])
print(y[::-1])
````

### List methods

- `append(value)` adds one item at the end
- `insert(index, value)` adds one item at that position
- `extend(other)` adds every item from another list
- `remove(value)` deletes the first matching value
- `pop()` deletes the last item; `pop(index)` deletes that position
- `count(value)` counts matches
- `sort()` sorts in place; `sort(reverse=True)` sorts high to low
- `reverse()` flips the current order
- `index(value)` returns the first position of that value
- `del items[i]` and `del items[start:stop]` delete by position
- `min`, `max`, and `len` report the smallest item, the largest item, and the length


````python
x = [1, 3, 56, 78, 3, 69]
x.append(45)
print("append", x)

x.insert(2, 77)
print("insert", x)

li = [1, 2, 3, 4, 5, 6, 7]
li.extend([8, 9, 10])
print("extend", li)

x.remove(78)
print("remove", x)

print("pop last", x.pop())
print("pop index 3", x.pop(3))
print("after pops", x)
print("count of 3", x.count(3))
````

````python
x = [1, 3, 3, 69, 77]
x.sort()
print("sorted", x)
x.sort(reverse=True)
print("sorted descending", x)

x = [1, 3, 56, 78, 3, 69]
x.reverse()
print("reversed", x)

vowels = ["a", "e", "i", "o", "u"]
print("index of i", vowels.index("i"))

x = [69, 3, 78, 56, 3, 1]
del x[2]
print("del one index", x)

x = [69, 3, 78, 56, 3, 1]
del x[2:6]
print("del a slice", x)
numbers = [69, 3, 56, 3, 1]
print("min", min(numbers), "max", max(numbers), "len", len(numbers))
````

## Unit 2 — Decisions and loops

Comparisons produce `True` or `False`:

- `==` equal
- `!=` not equal
- `<` `<=` `>` `>=`

An `if` runs its block only when the condition is true. `elif` tries another condition. `else` runs when earlier conditions were false.

The lines inside the block are indented, usually by four spaces. Python uses that indent to see where the block starts and ends.


````python
a = 33
b = 200
print(a == b)
print(a != b)
print(a < b)
print(a <= b)
print(a > b)
print(a >= b)

if b > a:
    print("b is greater than a")
````

````python
a = 33
b = 33
if b > a:
    print("b is greater than a")
elif a == b:
    print("a and b are equal")

a = 200
b = 33
if b > a:
    print("b is greater than a")
elif a == b:
    print("a and b are equal")
else:
    print("a is greater than b")

if b > a:
    print("b is greater than a")
else:
    print("b is not greater than a")
````

### One-line `if`, `and`, and `or`

A single action can sit on the same line as `if`. A value can also be chosen with `value_if_true if condition else value_if_false`.

`and` is true when both sides are true. `or` is true when at least one side is true.


````python
a = 200
b = 33
c = 500

if a > b: print("a is greater than b")

print("A") if a > b else print("B")
print("A") if a > b else print("=") if a == b else print("B")

if a > b and c > a:
    print("Both conditions are True")

if a > b or a > c:
    print("At least one of the conditions is True")
````

### `for` loops

`for item in sequence` runs once for each item in a list, tuple, string, set, or dictionary.

`print` ends with a newline. Pass `end=""` to stay on the same line.


````python
names = ["Sourabh", "Mahesh", "Pranay", "Arpit", "Lovelesh", "Anirudh"]
for name in names:
    print(name)

print("---")
for character in "Hello Programmers":
    print(character, end="")
print()
````

### `break` and `continue`

`break` leaves the loop. `continue` skips the rest of the current round.

Where you put `break` matters. After `print`, the matching item is shown. Before `print`, the loop stops without showing it.


````python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
    if fruit == "banana":
        break

print("---")
for fruit in fruits:
    if fruit == "banana":
        break
    print(fruit)

print("---")
for fruit in fruits:
    if fruit == "banana":
        continue
    print(fruit)
````

### `range`, `else` on a `for`, and nested loops

`range(stop)` yields `0` up to but not including `stop`. `range(start, stop)` starts at `start`. `range(start, stop, step)` jumps by `step`.

An `else` attached to a `for` runs after the loop finishes every item. A `break` skips that `else`.

A nested loop is a loop inside a loop. The inner loop runs completely once for each item of the outer loop.


````python
print("range(6)")
for n in range(6):
    print(n)

print("range(2, 6)")
for n in range(2, 6):
    print(n)

print("range(2, 30, 3)")
for n in range(2, 30, 3):
    print(n)

print("--- else after a finished loop ---")
for n in range(6):
    print(n)
else:
    print("Finally finished!")

print("--- else skipped after break ---")
for n in range(6):
    if n == 3:
        break
    print(n)
else:
    print("Finally finished!")
````

````python
adjectives = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]
for adjective in adjectives:
    for fruit in fruits:
        print(adjective, fruit)
````

### `while` loops

A `while` loop repeats as long as its condition stays true. Update the variable inside the loop. `break` and `continue` work here too.

In the `continue` example below, `i` increases at the start of the loop, so the printed values are `1, 2, 4, 5, 6`.


````python
i = 1
while i < 6:
    print(i)
    i += 1

print("--- break ---")
i = 1
while i < 6:
    print(i)
    if i == 3:
        break
    i += 1

print("--- continue ---")
i = 0
while i < 6:
    i += 1
    if i == 3:
        continue
    print(i)
````

## Unit 3 — Sets and tuples

### Sets

A set is unordered, so it has no index, and each value appears once. Write it with curly braces.

- loop with `for`, or test membership with `in`
- `add` inserts one item
- `update` inserts several items
- `len` counts items
- `remove` deletes a value and raises an error if it is missing
- `discard` deletes a value and does nothing if it is missing
- `pop` removes some item; because a set has no order, you do not choose which one
- `clear` empties the set
- `del` deletes the set name itself


````python
thisset = {"apple", "banana", "cherry"}
print(thisset)

for item in thisset:
    print(item)

print("banana" in thisset)

thisset.add("orange")
print("after add", thisset)

thisset.update(["orange", "mango", "grapes"])
print("after update", thisset)
print("length", len(thisset))
````

````python
thisset = {"apple", "banana", "cherry"}
thisset.remove("banana")
print("after remove", thisset)

thisset.discard("banana")
print("discard missing item", thisset)

thisset = {"apple", "banana", "cherry"}
removed = thisset.pop()
print("popped", removed)
print("remaining", thisset)

thisset.clear()
print("cleared", thisset)

thisset = {"apple", "banana", "cherry"}
del thisset
try:
    print(thisset)
except NameError as error:
    print("the name is gone:", type(error).__name__)
````

### Tuples

A tuple is ordered and unchangeable. Write it with parentheses. You can read an index, loop over it, test `in`, and ask for `len`.

Item assignment raises `TypeError`. `del` can delete the whole tuple. After that, the name no longer exists.


````python
thistuple = ("apple", "banana", "cherry")
print(thistuple)
print(thistuple[1])

for item in thistuple:
    print(item)

if "apple" in thistuple:
    print("Yes, 'apple' is in the fruits tuple")

print("length", len(thistuple))

try:
    thistuple[3] = "orange"
except TypeError as error:
    print("cannot change a tuple:", error)

del thistuple
try:
    print(thistuple)
except NameError as error:
    print("the tuple name is gone:", type(error).__name__)
````

## Unit 4 — Functions, classes, and objects

### Functions

`def` creates a function. The code runs when you call the name with parentheses. Parameters are the names inside the definition. A default value is used when the caller leaves that argument out.

A list passed in is still a list inside the function. `return` sends a value back.


````python
def my_function():
    print("Hello from a function")

my_function()


def greet(fname):
    print(fname + " Refsnes")

greet("Emil")
greet("Tobias")
greet("Linus")


def origin(country="Norway"):
    print("I am from " + country)

origin("Sweden")
origin("India")
origin()
origin("Brazil")
````

````python
def show_food(food):
    for item in food:
        print(item)

show_food(["apple", "banana", "cherry"])


def times_five(x):
    return 5 * x

print(times_five(3))
print(times_five(5))
print(times_five(9))
````

### Four shapes of a function

1. No input, no output — nothing comes in, and it only prints
2. No input, yes output — nothing comes in, and it `return`s a value
3. Yes input, no output — it receives arguments and prints
4. Yes input, yes output — it receives arguments and `return`s a value


````python
def say_ready():
    print("Ready")

def course_name():
    return "Python"

def show_total(a, b):
    print(a + b)

def add(a, b):
    return a + b

say_ready()
print(course_name())
show_total(4, 6)
print(add(4, 6))
````

### Classes and objects

A class is a blueprint. An object is one value built from that class.

A name assigned directly in the class is shared as a class attribute. `__init__` runs when you create an object and stores values on that object with `self`.

A method is a function written inside the class. Its first parameter refers to the object. The usual name is `self`. The language only requires that it is the first parameter.


````python
class MyClass:
    x = 5

p1 = MyClass()
print(p1.x)
````

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def myfunc(self):
        print("Hello my name is " + self.name)

p1 = Person("John", 36)
print(p1.name)
print(p1.age)
p1.myfunc()
````

````python
class Person:
    def __init__(mysillyobject, name, age):
        mysillyobject.name = name
        mysillyobject.age = age

    def myfunc(abc):
        print("Hello my name is " + abc.name)

p1 = Person("John", 36)
p1.myfunc()
````

### Change or delete a property, or delete the object

Assign a new value to an attribute to change it. `del object.attribute` removes that attribute. `del object` removes the object name. Using a deleted name raises `NameError`. Using a deleted attribute raises `AttributeError`.


````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("John", 36)
p1.age = 40
print("updated age", p1.age)

del p1.age
try:
    print(p1.age)
except AttributeError as error:
    print("age was deleted:", error)

del p1
try:
    print(p1)
except NameError as error:
    print("object name was deleted:", type(error).__name__)
````

### Opening a file

`open(filename, mode)` returns a file object. The mode is a short string:

| Mode | Meaning |
| --- | --- |
| `"r"` | read; error if the file is missing |
| `"a"` | append; create the file if needed |
| `"w"` | write; create the file, or wipe it if it already exists |
| `"x"` | create; error if the file already exists |
| `"t"` | text, the default |
| `"b"` | binary |

`"rt"` means read text. That is the default, so `open(path)` and `open(path, "rt")` do the same job.

Close the file when you finish. These cells write into a folder named `notes_files` next to the notebook.


````python
from pathlib import Path

folder = Path("notes_files")
folder.mkdir(exist_ok=True)

demo = folder / "demofile.txt"
f = open(demo, "w")
f.write("Hello! Welcome to demofile.txt\n")
f.write("This file is for testing purposes.\n")
f.write("Good Luck!\n")
f.close()

f = open(demo, "r")
print(f.read())
f.close()

f = open(demo, "r")
print("first 5 characters:", repr(f.read(5)))
f.close()

f = open(demo, "r")
print("first line:", repr(f.readline()))
print("second line:", repr(f.readline()))
f.close()

f = open(demo, "r")
for line in f:
    print(line, end="")
f.close()
````

### Exception handling

An exception stops the normal flow. `try` watches a block. `except` runs when a matching exception is raised. You can write several `except` blocks for different errors.

`else` runs when the `try` block raised nothing. `finally` runs afterward in either case. That is a useful place to close a file.


````python
try:
    print(not_defined_yet)
except NameError:
    print("Variable not_defined_yet is not defined")
except:
    print("Something else went wrong")
````

````python
try:
    print("Hello")
except NameError:
    print("Something went wrong")
else:
    print("Nothing went wrong")

print("---")
try:
    print(not_defined_yet)
except NameError:
    print("Something went wrong")
finally:
    print("The 'try except' is finished")
````

````python
from pathlib import Path

folder = Path("notes_files")
folder.mkdir(exist_ok=True)
path = folder / "demofile.txt"
f = open(path, "w")
f.write("Hello! Welcome to demofile.txt\n")
f.close()

f = open(path, "r")
try:
    f.write("Lorem Ipsum")
except OSError:
    print("Something went wrong when writing to the file")
finally:
    f.close()
    print("The file is closed. Closed flag:", f.closed)

print(path.read_text())
````
