## 1. Printing and comments

`print()` shows a value on the screen. Comments start with `#` and are ignored by Python. Use them to note why a line exists, not to repeat what the code already says.


````python
print("Hello, Python")
print(2 + 3)

# This line is a comment. Python skips it.
print("Comments are for people reading the code")
````

## 2. Variables and data types

A variable is a name for a value. You create one with `=`.

| Type | Example | What it stores |
| --- | --- | --- |
| `int` | `21` | whole number |
| `float` | `3.5` | number with a decimal |
| `str` | `"Asha"` | text |
| `bool` | `True` or `False` | yes / no |
| `None` | `None` | no value yet |

`type()` tells you which kind of value a variable holds.


````python
name = "Asha"
age = 21
height = 5.4
is_student = True
grade = None

print(name, type(name))
print(age, type(age))
print(height, type(height))
print(is_student, type(is_student))
print(grade, type(grade))

# A later line can give the same name a new value.
grade = "A"
print("grade is now", grade)
````

The cell below assigns `answer` directly so it runs without waiting for you to type. When you want to type a value yourself, uncomment the `input` line.

## 3. Operators

**Arithmetic:** `+` `-` `*` `/` `//` `%` `**`

- `/` always gives a float
- `//` is floor division (how many whole times)
- `%` is the remainder
- `**` is power

**Comparison** returns `True` or `False`: `==` `!=` `<` `>` `<=` `>=`

`=` stores a value. `==` checks whether two values are equal.

**Logical:** `and`, `or`, `not`


````python
a = 17
b = 5

print("sum", a + b)
print("difference", a - b)
print("product", a * b)
print("division", a / b)
print("floor division", a // b)
print("remainder", a % b)
print("power", b ** 3)

print("equal?", a == b)
print("greater?", a > b)
print("a is between 10 and 20?", a >= 10 and a <= 20)
print("not False:", not False)
````

## 4. Strings

Text lives in single quotes or double quotes. A string is a sequence of characters. Positions start at `0`. Negative positions count from the end.

```
 index:  0  1  2  3  4
 value:  h  e  l  l  o
        -5 -4 -3 -2 -1
```

Slicing uses `text[start:stop]`. `stop` is not included.


````python
message = "hello python"

print(message[0])
print(message[-1])
print(message[0:5])
print(message[6:])
print(len(message))

print(message.upper())
print(message.title())
print(message.replace("python", "world"))
print("python" in message)

words = message.split()
print(words)
print("-".join(words))
````

````python
name = "Ravi"
marks = 86

# An f-string inserts values inside the text.
print(f"{name} scored {marks}")
print(f"Next year {name} will be older, and 10 more marks would be {marks + 10}")
````

## 5. Lists

A list keeps several values in order. Square brackets create one. Lists can grow, shrink, and change.

Useful methods: `append`, `insert`, `pop`, `remove`, `sort`.


````python
marks = [72, 85, 64, 91]
print(marks[0])
print(marks[-1])
print(marks[1:3])

marks.append(78)
marks.insert(0, 60)
print("after adding:", marks)

last = marks.pop()
print("popped", last, "list is now", marks)

marks.remove(64)
print("after remove:", marks)

marks.sort()
print("sorted:", marks)
print("highest:", max(marks), "count:", len(marks))
````

## 6. Tuples

A tuple is an ordered collection that you do not change after creating it. Parentheses create one. Use a tuple for a fixed group of values, such as a point or a date.

You can unpack a tuple into separate variables.


````python
point = (3, 4)
print(point[0], point[1])
print(len(point))

x, y = point
print(f"x={x}, y={y}")

rgb = (255, 128, 0)
red, green, blue = rgb
print(red, green, blue)
````

## 7. Dictionaries

A dictionary stores pairs of keys and values. Curly braces create one. Look up a value by its key.

Use `.get(key, default)` when the key might be missing. A missing key with `[]` raises `KeyError`.


````python
student = {
    "name": "Neha",
    "roll": 23,
    "marks": 88,
}

print(student["name"])
print(student.get("city", "unknown"))

student["city"] = "Bhopal"
student["marks"] = 91
print(student)

for key, value in student.items():
    print(key, "->", value)
````

## 8. Sets

A set stores unique values, with no fixed order. Curly braces with only values (no `key: value` pairs) create a set. Adding a value that is already there does nothing.


````python
colors = {"red", "blue", "red", "green"}
print(colors)

colors.add("blue")
colors.add("yellow")
print("after add:", colors)
print("blue" in colors)

primary = {"red", "blue", "yellow"}
print("shared:", colors & primary)
print("either:", colors | primary)
````

## 9. Decisions: `if`, `elif`, `else`

Python runs one branch. It checks `if` first, then each `elif`, and uses `else` only when nothing earlier matched.

Indentation is part of the syntax. Lines inside a branch are indented by four spaces.


````python
marks = 76

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
elif marks >= 40:
    grade = "D"
else:
    grade = "F"

print(f"{marks} -> {grade}")

# Change marks to 35, 60, and 95, then run this cell again.
````

## 10. Loops

`for` walks through each item in a sequence. `range(n)` produces `0` through `n - 1`. `range(start, stop)` stops before `stop`.

`while` repeats as long as its condition is true. Something inside the loop must change, or it never ends.

- `break` leaves the loop immediately
- `continue` skips the rest of this round and starts the next one


````python
scores = [55, 82, 40, 91]

for score in scores:
    print("score:", score)

print("---")

for number in range(1, 5):
    print(number)

print("---")

# Build a new list from an old one.
passed = []
for score in scores:
    if score >= 40:
        passed.append(score)
print("passed:", passed)

# The same filter, written as a list comprehension.
passed_again = [score for score in scores if score >= 40]
print("passed again:", passed_again)
````

````python
total = 0
n = 1
while n <= 5:
    total = total + n
    n = n + 1
print("sum 1 through 5:", total)

print("---")

for n in range(1, 8):
    if n % 2 == 0:
        continue
    if n > 5:
        break
    print("odd and at most 5:", n)
````

## 11. Functions

A function is a named block of code you can run whenever you need it. `def` creates it. `return` sends a value back to the caller. Without `return`, the function gives back `None`.

- Parameters are the names in the definition.
- Arguments are the values you pass in the call.
- A default value is used when the caller omits that argument.


````python
def greet(name):
    print(f"Hello, {name}")

greet("Asha")
greet("Ravi")
````

````python
def average(numbers):
    """Return the mean of a non-empty list of numbers."""
    return sum(numbers) / len(numbers)

def letter_grade(marks, passing=40):
    if marks >= 90:
        return "A"
    if marks >= 75:
        return "B"
    if marks >= passing:
        return "C"
    return "F"

print(average([70, 80, 90]))
print(letter_grade(82))
print(letter_grade(35))
print(letter_grade(50, passing=60))
````

### Names inside and outside a function

A function can read a variable from outside. A name you assign inside the function is a new local variable. It does not change the outside name.


````python
college = "LNCT"

def show_place():
    room = "Lab 2"
    print(college, room)

show_place()
print("outside, college is still", college)

def rebind():
    college = "another campus"
    print("inside:", college)

rebind()
print("outside:", college)
````

## 12. Exception handling

When Python cannot finish a statement, it raises an exception and stops that cell. Common ones:

| Exception | Typical cause |
| --- | --- |
| `ZeroDivisionError` | divide by zero |
| `ValueError` | a value of the right type is still not acceptable, such as `int("abc")` |
| `TypeError` | an operation does not support that type |
| `IndexError` | a list index does not exist |
| `KeyError` | a dictionary key does not exist |

`try` / `except` lets you catch an exception and keep going. Name the exception you expect. A bare `except:` hides bugs, so avoid it.

- `else` runs when the `try` block did not raise
- `finally` runs whether or not an exception happened
- `raise` creates an exception yourself


````python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Cannot divide by zero")
        return None
    else:
        print("division worked")
        return result
    finally:
        print("safe_divide finished")

print("result:", safe_divide(10, 2))
print("---")
print("result:", safe_divide(10, 0))
````

````python
raw_values = ["12", "3.5", "hello", "9"]

for raw in raw_values:
    try:
        number = int(raw)
    except ValueError as error:
        print(f"skipped {raw!r}: {error}")
    else:
        print("whole number:", number)

def require_mark(mark):
    if mark < 0 or mark > 100:
        raise ValueError("mark must be between 0 and 100")
    return mark

try:
    print(require_mark(110))
except ValueError as error:
    print("rejected:", error)
````

## 13. Classes and objects

A class is a blueprint. An object is one thing built from that blueprint.

- `__init__` runs when you create an object and sets up its data
- `self` is the object itself. Use it to store attributes and call other methods
- A method is a function defined inside the class. Its first parameter is `self`
- Each object has its own attribute values

`__str__` controls what `print()` shows for the object.


````python
class Student:
    def __init__(self, name, roll):
        self.name = name
        self.roll = roll
        self.marks = []

    def add_mark(self, mark):
        if mark < 0 or mark > 100:
            raise ValueError("mark must be between 0 and 100")
        self.marks.append(mark)

    def average(self):
        if not self.marks:
            return 0.0
        return sum(self.marks) / len(self.marks)

    def __str__(self):
        return f"{self.name} (roll {self.roll}), average {self.average():.1f}"


asha = Student("Asha", 11)
asha.add_mark(80)
asha.add_mark(95)
print(asha)
print("marks:", asha.marks)
````

Two objects made from the same class do not share their lists or other attributes. Changing one student leaves the other student alone.


````python
ravi = Student("Ravi", 12)
neha = Student("Neha", 13)

ravi.add_mark(70)
neha.add_mark(90)
neha.add_mark(88)

print(ravi)
print(neha)
print("same marks list?", ravi.marks is neha.marks)
````

### Inheritance

A child class reuses a parent class and can add its own data. `super().__init__(...)` runs the parent setup first. This is the usual next step after one class and its objects.


````python
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        return f"Hello, I am {self.name}"


class CollegeStudent(Person):
    def __init__(self, name, roll):
        super().__init__(name)
        self.roll = roll

    def greet(self):
        return f"{super().greet()} and my roll number is {self.roll}"


person = Person("Kabir")
student = CollegeStudent("Kabir", 21)
print(person.greet())
print(student.greet())
````

## 14. Putting the pieces together

This example uses a class, a function, and exception handling. Bad marks are reported and skipped. Valid marks stay on the student.


````python
def record_marks(student, raw_marks):
    for mark in raw_marks:
        try:
            student.add_mark(mark)
        except (TypeError, ValueError) as error:
            print(f"skipped {mark!r}: {error}")


riya = Student("Riya", 18)
record_marks(riya, [88, 92, -5, 76, 110])
print(riya)
print("saved marks:", riya.marks)
````
